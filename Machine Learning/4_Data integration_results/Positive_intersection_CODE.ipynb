```python
# Seleccionar columnas
KNN_df = KNN_df[['ChemicalName', 'SMILES', 'Total Molweight', 'cLogP', 'H-Acceptors', 'H-Donors', 'Total Surface Area', 'Relative PSA', 'Rotatable Bonds', 'Predicciones']]
SVM_df = SVM_df[['ChemicalName', 'SMILES', 'Total Molweight', 'cLogP', 'H-Acceptors', 'H-Donors', 'Total Surface Area', 'Relative PSA', 'Rotatable Bonds', 'Prediction']]
tree_df = tree_df[['ChemicalName', 'SMILES', 'Total Molweight', 'cLogP', 'H-Acceptors', 'H-Donors', 'Total Surface Area', 'Relative PSA', 'Rotatable Bonds','Prediccion']]

# Cambiar nombre a columnas
KNN_df.columns = ['Name', 'SMILES', 'Total Molweight', 'cLogP', 'H-Acceptors', 'H-Donors', 'Total Surface Area', 'Relative PSA', 'Rotatable Bonds', 'Senotherapeutics:']
SVM_df.columns = ['Name', 'SMILES', 'Total Molweight', 'cLogP', 'H-Acceptors', 'H-Donors', 'Total Surface Area', 'Relative PSA', 'Rotatable Bonds', 'Senotherapeutics:']
tree_df.columns = ['Name', 'SMILES', 'Total Molweight', 'cLogP', 'H-Acceptors', 'H-Donors', 'Total Surface Area', 'Relative PSA', 'Rotatable Bonds', 'Senotherapeutics:']
```

```python
# positivos indicados en Senotherapeutics para cada dataframe

# Obtener los positivos para cada modelo
KNN_positivos = KNN_df[KNN_df['Senotherapeutics:'] == 1]['Name'].tolist()
SVM_positivos = SVM_df[SVM_df['Senotherapeutics:'] == 1]['Name'].tolist()
tree_positivos = tree_df[tree_df['Senotherapeutics:'] == 1]['Name'].tolist()

print("KNN Positivos:", len(KNN_positivos))
print("SVM Positivos:", len(SVM_positivos))
print("Árbol Positivos:", len(tree_positivos))
```

```python
# prompt: encuentra las filas que sean iguales en los tres dataframe

# Encontrar la intersección de los tres conjuntos de positivos
common_positives = set(KNN_positivos) & set(SVM_positivos) & set(tree_positivos)

print("Compuestos comunes en los tres modelos:", len(common_positives))

# Mostrar los nombres de los compuestos comunes
print(list(common_positives))


# Crear un nuevo dataframe con los compuestos comunes
common_df = pd.concat([
    KNN_df[KNN_df['Name'].isin(common_positives)],
    SVM_df[SVM_df['Name'].isin(common_positives)],
    tree_df[tree_df['Name'].isin(common_positives)]
])


```

```python
# Mostrar las filas que son iguales en los tres dataframes
common_df.sort_values('Name').drop_duplicates(subset='Name')
```
